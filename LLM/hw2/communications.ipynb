## Сравнение с PyTorch

`check.py` сравнивает результаты с PyTorch на трёх размерах частей.
Запустите обе ячейки после реализации функций и сохраните их вывод.


```python
import subprocess
import sys

def run_check(processes):
    result = subprocess.run(
        [sys.executable, "-m", "torch.distributed.run", "--standalone", f"--nproc-per-node={processes}", "check.py"],
        capture_output=True, text=True,
    )
    print(result.stdout, end="")
    if result.stderr:
        print(result.stderr, end="", file=sys.stderr)
    result.check_returncode()
```

```python
run_check(2)
```

```python
run_check(4)
```
